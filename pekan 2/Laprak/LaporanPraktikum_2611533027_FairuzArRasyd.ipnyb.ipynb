<h1 align="center"><b>BAB II<br>PEMBAHASAN</b></h1>

## 2.1 Dasar Teori

1. **Integer (int)**  
Integer adalah tipe data yang digunakan untuk menyimpan bilangan bulat, baik positif, negatif, maupun nol.

2. **Float**  
Float digunakan untuk menyimpan bilangan yang memiliki nilai desimal, misalnya 10.5 atau 3.14.

3. **Complex**  
Complex digunakan untuk menyimpan bilangan kompleks yang memiliki bagian real dan imajiner, misalnya `100+3j`.

4. **Boolean**  
Boolean merupakan tipe data yang memiliki dua nilai, yaitu `True` dan `False`. Tipe data ini sering digunakan untuk menentukan kondisi dalam program.

5. **String**  
String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String ditulis menggunakan tanda kutip.

6. **Konstanta**  
Konstanta merupakan nilai yang dimaksudkan untuk tetap selama program dijalankan. Pada Python, penamaan dengan huruf kapital sering digunakan sebagai konvensi untuk menandai konstanta.

7. **Input**  
Fungsi `input()` digunakan untuk menerima data yang dimasukkan oleh pengguna. Data hasil `input()` pada dasarnya berupa string sehingga dapat dikonversi ke tipe data lain jika diperlukan.

### 2.2.2 Program "Numerik_2611533027"
</div>

```python
# Buat file dengan nama Numerik_2611533027.py

a=100
print("The type of variable having value:", a, "is", type(a))

b=10.2345
print("The type of variable having value:", b, "is", type(b))

c=100+3j
print("The type of variable having value:", c, "is", type(c))
```

### 2.2.3 Program "Boolean_2611533027"
</div>

```python
# Buat file dengan nama Boolean_2611533027.py

is_lulus = True
is_cumlaude = True

nilai = 85
batas_lulus = 75

status_kelulusan = nilai >= batas_lulus

print("=== Check Kelulusan ===")
print("Nilai:", nilai)
print("Apakah lulus?:", status_kelulusan)

if is_lulus and is_cumlaude:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

<div style="font-family:'Times New Roman', Times, serif; font-size:18px;">
**Penjelasan:**

1. `is_lulus_3027` dan `is_cumlaude_3027` menyimpan nilai boolean.
2. `nilai_3027` menyimpan nilai 85 sebagai integer.
3. `batas_lulus_3027` menyimpan batas kelulusan sebesar 75.
4. Perbandingan `nilai_3027 >= batas_lulus_3027` menghasilkan nilai `True`.
5. Percabangan `if` digunakan untuk menampilkan pesan ketika kedua kondisi bernilai benar.
</div>

<div style="font-family:'Times New Roman', Times, serif; font-size:18px;">

### 2.2.4 Program "Input_2611533027"
</div>

```python
# Buat file dengan nama Input_2611533027.py
# Program menghitung luas persegi panjang

panjang_3027 = input("Masukkan nilai panjang: ")
lebar_3027 = input("Masukkan nilai lebar: ")

luas_3027 = int(panjang_3027) * int(lebar_3027)

print("Luas:", luas_3027)
```

<div style="font-family:'Times New Roman', Times, serif; font-size:18px;">
**Penjelasan:**

1. `input()` digunakan untuk meminta nilai panjang dari pengguna.
2. `input()` digunakan kembali untuk meminta nilai lebar.
3. `int()` mengubah masukan menjadi bilangan integer.
4. Nilai panjang dan lebar dikalikan untuk mendapatkan luas persegi panjang.
5. Hasil perhitungan ditampilkan menggunakan `print()`.
</div>

<div style="font-family:'Times New Roman', Times, serif; font-size:18px;">

### 2.2.5 Program "Konstanta_2611533027"
</div>

```python
# Buat file dengan nama Konstanta_2611533027.py
# Program menggunakan konstanta untuk menghitung luas lingkaran

from typing import Final 
PI: Final = 3.14
print("pi: %f" (PI))
jari_3027 = float(input('Masukkan nilai jari-jari : '))
luas_3027 = PI * jari_3027 * jari_3027
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3027, luas_3027))
```

<div style="font-family:'Times New Roman', Times, serif; font-size:18px;">
**Penjelasan:**

1. `Final` diimpor dari library `typing`.
2. `PI: Final = 3.14` digunakan untuk mendeklarasikan nilai konstanta PI.
3. Fungsi `input()` digunakan untuk menerima nilai jari-jari.
4. `float()` mengubah masukan menjadi bilangan desimal.
5. Luas lingkaran dihitung dengan rumus `PI × r × r`.
6. Hasil perhitungan ditampilkan dengan dua angka di belakang koma.
