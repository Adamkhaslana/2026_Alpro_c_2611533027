### 2.1.1 Python

Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami sehingga cocok digunakan untuk mempelajari dasar-dasar pemrograman. ### 2.1.2 Visual Studio Code

Visual Studio Code atau VS Code merupakan text editor yang digunakan untuk menulis, mengedit, dan menjalankan kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python, serta menyediakan fitur seperti syntax highlighting, debugging, terminal, dan berbagai extension yang dapat membantu proses pemrograman. ### 2.1.3 Git

Git merupakan sistem pengontrol versi atau version control system yang digunakan untuk mencatat dan mengelola perubahan pada file dalam suatu proyek. Dengan Git, setiap perubahan yang dilakukan pada program dapat disimpan dalam bentuk riwayat sehingga pengguna dapat mengetahui perkembangan proyek dan mengembalikan file ke versi sebelumnya jika diperlukan. Git juga membantu dalam proses pengelolaan proyek, terutama ketika bekerja dengan banyak file atau bekerja secara bersama-sama dengan pengguna lain.
### 2.1.4 GitHub

GitHub merupakan platform berbasis internet yang digunakan untuk menyimpan, mengelola, dan membagikan proyek pemrograman menggunakan Git. GitHub memungkinkan pengguna menyimpan kode secara online sehingga proyek dapat diakses dari berbagai perangkat dan dapat digunakan untuk melakukan kolaborasi dengan pengguna lain. Dalam praktikum ini, GitHub digunakan sebagai tempat penyimpanan repository proyek Python yang dibuat melalui VS Code sehingga hasil pekerjaan dapat tersimpan secara terorganisir dan terhubung dengan sistem version control.
### 2.1.5 Repository

Repository atau repositori merupakan tempat penyimpanan proyek yang berisi file-file program beserta riwayat perubahan yang dikelola menggunakan Git. Repository dapat dibuat secara lokal pada komputer maupun secara online melalui GitHub. Dalam praktikum ini, repository digunakan untuk menyimpan file program Python dan berbagai file pendukung lainnya. Dengan adanya repository, seluruh perkembangan proyek dapat dicatat sehingga perubahan yang dilakukan dapat dipantau dan dikelola dengan lebih mudah.

## 2.3 Koneksi VS Code dengan GitHub

Setelah program selesai dibuat dan diuji melalui Visual Studio Code, proyek dapat dihubungkan dengan repository GitHub.

Langkah-langkah yang dilakukan:

1. Membuat repository baru pada GitHub.
2. Membuka folder proyek menggunakan Visual Studio Code.
3. Membuka Terminal pada Visual Studio Code.
4. Menginisialisasi repository Git.
5. Menambahkan file proyek ke staging area.
6. Membuat commit.
7. Menghubungkan repository lokal dengan repository GitHub.
8. Melakukan push ke GitHub.


### 2.3.1 Perintah Git

Jalankan perintah berikut pada Terminal VS Code:


````python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/USERNAME/NAMA-REPOSITORY.git
git push -u origin main
````

### Penjelasan Perintah

- `git init` digunakan untuk membuat repository Git pada folder proyek.
- `git add .` digunakan untuk menambahkan file ke staging area.
- `git commit` digunakan untuk mencatat perubahan.
- `git branch -M main` digunakan untuk menentukan branch utama.
- `git remote add origin` digunakan untuk menghubungkan repository lokal dengan GitHub.
- `git push` digunakan untuk mengunggah kode ke GitHub.

> **Catatan:** Perintah Git di atas dijalankan pada **Terminal VS Code**, bukan pada Code Cell Python.
