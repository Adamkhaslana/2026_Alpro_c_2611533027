## 2.1 Dasar Teori
1. Operator Aritmatika: Simbol matematika yang digunakan untuk melakukan operasi hitung dasar, seperti penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), pembagian bulat (`//`), sisa bagi/modulus (`%`), dan perpangkatan (`**`).

2. Operator Perbandingan (Relasional): Operator yang membandingkan dua nilai dan menghasilkan nilai boolean (`True` atau `False`). Meliputi lebih besar (`>`), lebih kecil (`<`), lebih besar sama dengan (`>=`), lebih kecil sama dengan (`<=`), sama dengan (`==`), dan tidak sama dengan (`!=`).

3. Operator Logika: Digunakan untuk membandingkan dua ekspresi boolean. Terdiri dari `and` (bernilai `True` jika keduanya `True`), `or` (bernilai `True` jika salah satu `True`), dan `not` (membalikkan nilai boolean).

4. Operator Assignment (Penugasan): Digunakan untuk memberikan atau menguji nilai ke dalam variabel. Operator gabungan (seperti `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`) melakukan operasi matematika sekaligus memperbarui nilai variabel tersebut.

5. Operator Keanggotaan & Identitas:

    - Keanggotaan (`in`, `not in`): Memeriksa apakah suatu nilai ada di dalam sebuah urutan/koleksi data (seperti List atau String).

    - Identitas (`is`, `is not`): Membandingkan apakah dua variabel merujuk pada objek/lokasi memori yang sama (berbeda dengan == yang hanya membandingkan nilainya).

6. Operator Bitwise: Operator yang memanipulasi bit angka secara langsung pada tingkat biner (0 dan 1). Terdiri dari AND (`&`), OR (`|`), XOR (`^`), NOT (`~`), Left Shift (`<<`), dan Right Shift (`>>`).


## 2.2 Kode Program

### 2.2.1 Program Aritmatika (`Aritmatika_2611533027.py`)
</div>

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3027 = int(input("Input angka-1: "))
angka2_3027 = int(input("Input angka-2: "))

# Penjumlahan
hasil_3027 = angka1_3027 + angka2_3027
print("\nOperator Penjumlahan:")
print("Hasil =", hasil_3027)

# Pengurangan
hasil_3027 = angka1_3027 - angka2_3027
print("\nOperator Pengurangan:")
print("Hasil =", hasil_3027)

# Perkalian
hasil_3027 = angka1_3027 * angka2_3027
print("\nOperator Perkalian:")
print("Hasil =", hasil_3027)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_3027 != 0:
    hasil_3027 = angka1_3027 / angka2_3027
    print("\nOperator Pembagian:")
    print("Hasil =", hasil_3027)

    hasil_3027 = angka1_3027 // angka2_3027
    print("\nOperator Pembagian Bulat:")
    print("Hasil =", hasil_3027)

    hasil_3027 = angka1_3027 % angka2_3027
    print("\nOperator Sisa Bagi:")
    print("Hasil =", hasil_3027)
else:
    print("Angka kedua tidak boleh bernilai 0.")

hasil_3027 = angka1_3027 ** angka2_3027
print("\nOperator Pangkat:")
print("Hasil =", hasil_3027)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

<b>Penjelasan:</b>

Program pada bagian ini digunakan untuk menguji operator aritmatika dengan dua bilangan yang dimasukkan pengguna. Nilai input terlebih dahulu diubah menjadi integer sehingga dapat diproses sebagai angka. Selanjutnya setiap operator, seperti penjumlahan, pengurangan, perkalian, pembagian, pembagian bulat, modulus, dan perpangkatan, diterapkan untuk melihat hasil perhitungannya.

Operator pembagian dilengkapi pemeriksaan terhadap nilai pembagi agar program tidak melakukan pembagian dengan nol. Dengan demikian, bagian ini tidak hanya memperlihatkan cara kerja operator matematika dalam Python, tetapi juga menunjukkan pentingnya validasi input sebelum suatu operasi dijalankan.

### 2.2.1 Program Perbandingan (`Perbandingan_2611533027.py`)
</div>


```python
# Buat file dengan nama Perbandingan_NIM.py
# Nama variabel ditambah 4 digit NIM terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program untuk operator Perbandingan dalam Python

angka1_3027 = int(input("Input angka-1: "))
angka2_3027 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_3027 = angka1_3027 > angka2_3027
print("\nOperator Lebih Besar Dari:")
print("angka1_3027 > angka2_3027:", hasil_3027)

# Lebih kecil dari
hasil_3027 = angka1_3027 < angka2_3027
print("\nOperator Lebih Kecil Dari:")
print("angka1_3027 < angka2_3027:", hasil_3027)

# lebih besar dari atau sama dengan
hasil_3027 = angka1_3027 >= angka2_3027
print("\nOperator Lebih Besar Dari Sama Dengan:")
print("angka1_3027 >= angka2_3027:", hasil_3027)

# lebih kecil dari atau sama dengan
hasil_3027 = angka1_3027 <= angka2_3027
print("\nOperator Lebih Kecil Dari Sama Dengan:")
print("angka1_3027 <= angka2_3027:", hasil_3027)

# Sama dengan
hasil_3027 = angka1_3027 == angka2_3027
print("\nOperator Sama Dengan:")
print("angka1_3027 == angka2_3027:", hasil_3027)

# Tidak sama dengan
hasil_3027 = angka1_3027 != angka2_3027
print("\nOperator Tidak Sama Dengan:")
print("angka1_3027 != angka2_3027:", hasil_3027)

#Tambahan: perbandingan berantai dalam Python
hasil_3027 = 0 < angka1_3027 < 100
print("\nPerbandingan berantai:")
print("0 < angka1_3027 < 100:", hasil_3027)

hasil_3027 = 0 < angka2_3027 < 100
print("0 < angka2_3027 < 100:", hasil_3027)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

<b>Penjelasan:</b>

Pada program perbandingan, dua bilangan dibaca dari input pengguna kemudian diuji menggunakan beberapa operator pembanding. Operator `>`, `<`, `>=`, `<=`, `==`, dan `!=` menghasilkan nilai boolean berdasarkan hubungan antara kedua angka. Selain itu, program juga menggunakan perbandingan berantai untuk memeriksa apakah sebuah angka berada dalam rentang tertentu.

Hasil setiap pemeriksaan disimpan ke variabel `hasil_3027` sebelum ditampilkan. Penggunaan operator tersebut membantu program menentukan hubungan dua nilai secara langsung dan memperlihatkan bahwa hasil perbandingan dalam Python berupa `True` atau `False`.

### 2.2.3 Program Logika (`Logika_2611533027.py`)
</div>


```python
# Masukkan nilai boolean
#input tidak peka terhadap huruf besar dan kecil
a1_3027 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3027 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1_3027 =", a1_3027)
print("A2_3027 =", a2_3027)

# Konjungsi: bernilai True jika keduanya True
hasil_3027 = a1_3027 and a2_3027
print("\nOperator Konjungsi (AND):")
print("A1_3027 and A2_3027:", hasil_3027)

# Disjungsi: bernilai True jika salah satunya True
hasil_3027 = a1_3027 or a2_3027
print("\nDisjungsi (OR)")
print("A1_3027 or A2_3027:", hasil_3027)

# Negasi A1: membalik nilai A1
hasil_3027 = not a1_3027
print("\nNegasi A1_3027 (NOT)")
print("not A1_3027:", hasil_3027)

# Negasi A2: membalik nilai A2
hasil_3027 = not a2_3027
print("\nNegasi A2_3027 (NOT)")
print("not A2_3027:", hasil_3027)

# XOR: bernilai True jika kedua nilai berbeda
hasil_3027 = a1_3027 != a2_3027
print("\nDisjungsi Eksklusif (XOR)")
print("A1_3027 XOR A2_3027:", hasil_3027)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

<b>Penjelasan:</b>

Bagian ini mengolah dua masukan yang merepresentasikan nilai boolean. Teks input dibersihkan menggunakan `.strip()` dan diseragamkan dengan `.lower()`, kemudian dibandingkan dengan `"true"` agar menghasilkan tipe boolean.

Setelah memperoleh kedua nilai tersebut, program mencoba beberapa operasi logika. `and` membutuhkan kedua kondisi bernilai benar, `or` cukup membutuhkan salah satu kondisi benar, sedangkan `not` membalik keadaan boolean. Operasi XOR direpresentasikan melalui `!=`, sehingga bernilai benar ketika kedua input memiliki nilai yang berbeda.

### 2.2.4 Program Assigment (`Assigment_2611533027.py`)
</div>


```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_3027 = int(input("Input angka-1: "))
angka2_3027 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_3027)
print("Nilai awal angka2 =", angka2_3027)

# Assigment biasa
hasil_3027 = angka1_3027
print("\nHasil assigment biasa (=)")
print("Hasil =", hasil_3027)

# Assigment penambahan
hasil_3027 = angka1_3027 
hasil_3027 += angka2_3027
print("\nAssigment penambahan (+=)")
print("Hasil =", hasil_3027)

# Assigment pengurangan
hasil_3027 = angka1_3027
hasil_3027 -= angka2_3027
print("\nAssigment pengurangan (-=)")
print("Hasil =", hasil_3027)

# Assigment perkalian
hasil_3027 = angka1_3027
hasil_3027 *= angka2_3027
print("\nAssigment perkalian (*=)")
print("Hasil =", hasil_3027)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_3027 != 0:
    hasil_3027 = angka1_3027
    hasil_3027 /= angka2_3027
    print("\nAssigment pembagian (/=)")
    print("Hasil =", hasil_3027)
    # Operator tambahan
    hasil_3027 = angka1_3027
    hasil_3027 //= angka2_3027
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =", hasil_3027)

    hasil_3027 = angka1_3027
    hasil_3027 %= angka2_3027
    print("\nAssigment sisa bagi (%=)")
    print("Hasil =", hasil_3027)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_3027 = angka1_3027
hasil_3027 **= angka2_3027
print("\nAssigment perpangkatan (**=)")
print("Hasil =", hasil_3027)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

<b>Penjelasan:</b>

Program assignment memperlihatkan bagaimana sebuah nilai dapat diberikan dan diperbarui menggunakan operator penugasan. Dua bilangan diterima sebagai integer, kemudian nilai pertama dipakai sebagai dasar untuk beberapa operasi assignment.

Selain operator `=`, program menerapkan `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=`. Khusus operasi pembagian, terdapat kondisi yang memastikan pembagi tidak bernilai nol. Cara ini membuat perubahan nilai pada variabel dapat diamati secara bertahap tanpa harus menuliskan bentuk operasi yang panjang.

### 2.2.5 Program Lainnya (`Lainnya_2611533027.py`)
</div>


```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("======================================")
print("1. OPERATOR KEANGGOTAAN")
print("======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3027 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3027 = [int(angka.strip()) for angka in input_data_3027.split(',')]

nilai_dicari_3027 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3027 = nilai_dicari_3027 in data_3027
print("\nOperator keanggotaan IN")
print(nilai_dicari_3027,"in",data_3027,"=",hasil_3027)

# Operator not in
hasil_3027 = nilai_dicari_3027 not in data_3027
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_3027,"not in",data_3027,"=",hasil_3027)

print("======================================")
print("2. OPERATOR IDENTIAS")
print("======================================")

# objek1 menggunakan list dari input perngguna
object1_3027 = data_3027

# objek2 merujuk pada objek yang sama dengan objek1
object2_3027 = object1_3027

# objek3 memiliki isi sama, tetapi merupakan objek baru
object3_3027 = data_3027.copy()

print("object1 =",object1_3027)
print("object2 =",object2_3027)
print("object3 =",object3_3027)

# Operator is
hasil_3027 = object1_3027 is object2_3027
print("\nOperator identitas IS")
print("objek1 is objek2 =",hasil_3027)

# Operator is not
hasil_3027 = object1_3027 is not object2_3027
print("\nOperator identitas IS NOT")
print("objek1 is not objek2 =",hasil_3027)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =",object1_3027 is object3_3027)
print("objek1 == objek3 =",object1_3027 == object3_3027)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

<b>Penjelasan:</b>

Program ini membahas dua kelompok operator, yaitu operator keanggotaan dan operator identitas. Data angka dimasukkan dalam satu baris dengan pemisah koma, lalu dipisahkan menggunakan `split()` dan dikonversi menjadi integer sehingga membentuk sebuah list.

Operator `in` dan `not in` digunakan untuk mengetahui keberadaan suatu nilai di dalam list. Selanjutnya `object2_3027` dibuat sebagai referensi ke objek yang sama dengan `object1_3027`, sedangkan `object3_3027` merupakan salinan baru melalui `copy()`. Perbedaan tersebut kemudian diperiksa menggunakan `is`, `is not`, dan perbandingan nilai sehingga terlihat bahwa kesamaan isi tidak selalu berarti objek yang sama.

### 2.2.6 Program Bitwise (`Bitwise_2611533027.py`)
</div>


```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_3027 = int(input("Masukkan angka bitwise-1: "))
angka2_3027 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_3027,"| biner =",bin(angka1_3027))
print("angka1 =",angka2_3027,"| biner =",bin(angka2_3027))

# Bitwise AND
hasil_3027 = angka1_3027 & angka2_3027
print("\nBitwise AND (&)")
print(angka1_3027,"&",angka2_3027,hasil_3027)
print("Biner hasil =",bin(hasil_3027))
print("Biner hasil (8 bit) =",format(hasil_3027,"08b"))

# Bitwise OR
hasil_3027 = angka1_3027 | angka2_3027
print("\nBitwise OR (|)")
print(angka1_3027,"|",angka2_3027,hasil_3027)
print("Biner hasil =",bin(hasil_3027))
print("Biner hasil (8 bit) =",format(hasil_3027,"08b"))

# Bitwise XOR
hasil_3027 = angka1_3027 ^ angka2_3027
print("\nBitwise XOR (^)")
print(angka1_3027,"^",angka2_3027,hasil_3027)
print("Biner hasil =",bin(hasil_3027))
print("Biner hasil (8 bit) =",format(hasil_3027,"08b"))

# Bitwise NOT
hasil_3027 = ~angka1_3027
print("\nBitwise NOT (~)")
print(angka1_3027,"~",angka2_3027,hasil_3027)
print("Biner hasil =",bin(hasil_3027))
print("Biner hasil (8 bit) =",format(hasil_3027,"08b"))

# Bitwise geser kiri
jumlah_geser_3027 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3027 = angka1_3027 << jumlah_geser_3027
print("\nBitwise geser kiri (<<)")
print(angka1_3027,"<<",jumlah_geser_3027,"=",hasil_3027)
print("Biner hasil =",bin(hasil_3027))
print("Biner hasil (8 bit) =",format(hasil_3027,"08b"))

# Bitwise geser kanan
hasil_3027 = angka1_3027 >> jumlah_geser_3027
print("\nBitwise geser kanan (>>)")
print(angka1_3027,">>",jumlah_geser_3027,"=",hasil_3027)
print("Biner hasil =",bin(hasil_3027))
print("Biner hasil (8 bit) =",format(hasil_3027,"08b"))
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

<b>Penjelasan:</b>

Program bitwise menerima dua bilangan desimal dan menampilkan bentuk binernya menggunakan `bin()`. Setelah itu, operasi dilakukan langsung terhadap representasi bit melalui operator `&`, `|`, `^`, dan `~`. Hasilnya juga ditampilkan dalam bentuk biner agar perubahan setiap bit lebih mudah diamati.

Program kemudian menggunakan operator geser `<<` dan `>>`. Pergeseran kiri memindahkan bit ke posisi yang lebih tinggi, sedangkan pergeseran kanan memindahkannya ke posisi yang lebih rendah. Jumlah pergeseran ditentukan oleh pengguna sehingga hasil operasi dapat diuji dengan beberapa nilai berbeda.

```python
print("=== SISTEM TRANSAKSI TOKO ===")
nama_3027 = input("Masukkan Nama Pelanggan : ")
status_3027 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_3027 = int(input("Masukkan Total Belanja : "))
jumlah_3027 = int(input("Masukkan Jumlah Barang : "))
promo_3027 = input("Masukkan Kode Promo : ")

print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan       : {nama_3027}")
print(f"Status Pelanggan     : {status_3027}")
print(f"Total Belanja        : Rp{total_3027}")
print(f"Jumlah Barang        : {jumlah_3027}")
print(f"Kode Promo           : {promo_3027}")

kode_promo_3027 = ["HEMAT10", "HEMAT20", "HIDUPMAHASISWA", "HIDUPFTI"] #list kode promo

syarat_total_3027 = total_3027 >= 200000 # apakah memenuhi syarat belanja
syarat_jumlah_3027 = jumlah_3027 >= 3 # apakah memenuhi syarat barang
status_valid_3027 = status_3027 == "member" # apakah user member
promo_valid_3027 = promo_3027 in kode_promo_3027 # apakah kode promo ada dalam list

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_total_3027}")
print(f"Jumlah Barang >= 3         : {syarat_jumlah_3027}")
print(f"Status Member              : {status_valid_3027}")
print(f"Kode Promo Tersedia        : {promo_valid_3027}")
print(f"Mendapatkan Diskon         : {syarat_jumlah_3027 or syarat_total_3027}")
print(f"Mendapatkan Promo          : {promo_valid_3027}")

diskon_3027 = 0.05

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                      : Rp{diskon_3027 * total_3027}")
print(f"Total Pembayaran            : Rp{total_3027 - total_3027 * diskon_3027}")
print(f"Rata-rata Harga Barang      : Rp{(total_3027 - total_3027 * diskon_3027)/jumlah_3027}")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses              : ...")
print(f"Member Access               : {status_valid_3027}")
print(f"Promo Access                : {promo_valid_3027}")
print(f"Free Shipping Access        : ...")

kode_transaksi_3027 = int(status_valid_3027) << 0 | int(syarat_total_3027) << 1 | int(syarat_jumlah_3027) << 2 | int(promo_valid_3027) << 3
kode_referensi_3027 = int(status_valid_3027) << 0 | int(syarat_total_3027) << 1 | int(promo_valid_3027) << 3

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print(f"{format(int(status_valid_3027) << 0,'04b')} | {format(int(syarat_total_3027) << 1,'04b')} | {format(int(syarat_jumlah_3027) << 2,'04b')} | {format(int(promo_valid_3027) << 3,'04b')}")
print(f"Kode Biner   : {format(kode_transaksi_3027,'04b')}")
print(f"Kode Desimal : {kode_transaksi_3027}")

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{format(kode_transaksi_3027,'04b')} & {format(int(status_valid_3027) << 0,'04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_3027) & int(status_valid_3027) << 0,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3027) & int(status_valid_3027) << 0}")

print("Cek Promo")
print(f"{format(kode_transaksi_3027,'04b')} & {format(int(promo_valid_3027) << 3,'04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_3027) & int(promo_valid_3027) << 3,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3027) & int(promo_valid_3027) << 3}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi : {format(kode_transaksi_3027,'04b')}")
print(f"Kode Referensi : {format(kode_referensi_3027,'04b')}")
print(f"{format(kode_transaksi_3027,'04b')} ^ {format(kode_referensi_3027,'04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_3027) ^ (kode_referensi_3027),'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3027) ^ (kode_referensi_3027)}")

print("\n=== Shift ===")
print(f"{format(kode_transaksi_3027,'04b')} << 1")
print(f"Hasil Biner   : {format((kode_transaksi_3027) << 1,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3027) << 1}") 
print("=== SELESAI ===")
```
