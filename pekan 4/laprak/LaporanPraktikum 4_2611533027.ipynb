## 2.1 Dasar Teori
1. <b>If Tunggal:</b> Bentuk percabangan paling sederhana. Perintah di dalamnya hanya dijalankan ketika syarat yang diperiksa menghasilkan nilai `True`.

2. <b>If-Else:</b> Memiliki dua kemungkinan jalur. Program menjalankan bagian `if` ketika kondisi benar, sedangkan bagian `else` menjadi pilihan ketika kondisi tersebut tidak terpenuhi.

3. <b>If-Elif-Else:</b> Digunakan ketika terdapat beberapa kemungkinan kondisi. Python memeriksa kondisi dari bagian paling atas dan berhenti pada cabang pertama yang menghasilkan `True`.

4. <b>Multi-IF Terpisah:</b> Terdiri dari beberapa perintah `if` yang diperiksa secara mandiri. Karena tidak saling terikat dengan `elif`, lebih dari satu kondisi dapat menghasilkan aksi dalam satu proses.

5. <b>Match-Case:</b> Struktur pencocokan pola yang tersedia sejak Python 3.10. Struktur ini berguna ketika sebuah nilai perlu dibandingkan dengan sejumlah pilihan tertentu, sedangkan `case _` digunakan untuk menangani pilihan yang tidak cocok.

## 2.2 Kode Program
### 2.2.1 Program IF 1 (`if1_2611533027.py`)

</div>

````python
# Nama file program menggunakan format if1_nim.py
# Program menerapkan percabangan kondisi
# Penamaan variabel menggunakan empat digit terakhir NIM
# Data pengguna dibaca melalui fungsi input()

ipk_3027 = float(input("Input IPK Anda = "))

if ipk_3027 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3027))

print("Program Selesai")
````

<div style="font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

<b>Penjelasan:</b><br><br>
<b>1. Mengambil Nilai IPK</b><br>
`ipk_3027 = float(input("Input IPK Anda = "))`<br>
Baris tersebut menampilkan permintaan nilai IPK. Data dari `input()` pada awalnya berbentuk teks, kemudian `float()` digunakan agar nilai dapat diproses sebagai bilangan desimal.

<br><br>
<b>2. Pemeriksaan dengan IF</b><br>
`if ipk_3027 > 2.75:`<br>
&emsp;`print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3027))`<br>
Kondisi memeriksa apakah nilai IPK melewati angka 2.75. Jika syarat terpenuhi, program menampilkan informasi kelulusan. Apabila syarat tidak terpenuhi, bagian di dalam `if` tidak dijalankan dan program langsung menuju perintah berikutnya.

### 2.2.2 Program IF 2 (`if2_2611533027.py`)

</div>

````python
# Nama file program menggunakan format if2_nim.py
# Program menerapkan percabangan kondisi
# Penamaan variabel menggunakan empat digit terakhir NIM
# Data pengguna dibaca melalui fungsi input()

ipk_3027 = float(input("Input IPK Anda = "))

if ipk_3027 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3027))
else:  
    print("Anda Tidak Lulus")
print("Program Selesai")
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

<b>Penjelasan:</b><br><br>
<b>1. Membaca Input IPK</b><br>
`ipk_3027 = float(input("Input IPK Anda = "))`<br>
Pengguna memasukkan nilai IPK, kemudian masukan tersebut diubah menjadi tipe `float` agar dapat dibandingkan dengan batas nilai yang telah ditentukan.

<br><br>
<b>2. Menentukan Dua Kemungkinan Hasil</b><br>
`if ipk_3027 > 2.75:`<br>
`else:`<br>
Program mempunyai dua jalur keluaran. Nilai di atas 2.75 akan diproses oleh bagian `if`, sedangkan nilai 2.75 atau lebih rendah akan membuat program menjalankan bagian `else` dan menampilkan informasi bahwa pengguna tidak lulus.

### 2.2.3 Program Multi-IF 1 (`multi_if1_2611533027.py`)

</div>

````python
# Nama file program menggunakan format multi_if1_nim.py
# Program menerapkan percabangan kondisi
# Penamaan variabel menggunakan empat digit terakhir NIM
# Data pengguna dibaca melalui fungsi input()

umur_3027 = int(input("Input Umur Anda: "))
sim_3027 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_3027 >= 17 and sim_3027 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")
if umur_3027 >= 17 and sim_3027 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
if umur_3027 < 17 and sim_3027 == "y":
    print("Anda Belum Cukup Umur punya SIM")
if umur_3027 < 17 and sim_3027 != "y":
    print("Anda Belum Cukup Umur bawa motor")
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

<b>Penjelasan:</b><br><br>
<b>1. Pemeriksaan Kondisi Secara Terpisah</b><br>
`if umur_3027 >= 17 and sim_3027 == "y":`<br>
`if umur_3027 >= 17 and sim_3027 != "y":`<br>
`if umur_3027 < 17 and sim_3027 == "y":`<br>
`if umur_3027 < 17 and sim_3027 != "y":`<br><br>
Keempat kondisi tersebut tidak dibuat dalam satu rangkaian `if-elif`. Masing-masing diperiksa oleh Python secara independen dengan menggabungkan umur dan status SIM menggunakan operator `and`.

</div>

### 2.2.4 Program IF-ELIF-ELSE 1 (`if_elif_else1_2611533027.py`)

</div>

````python
# Nama file program menggunakan format if_elif_else1_nim.py
# Program menerapkan percabangan kondisi
# Penamaan variabel menggunakan empat digit terakhir NIM
# Data pengguna dibaca melalui fungsi input()

umur_3027 = int(input("Input Umur Anda: "))
sim_3027 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_3027 >= 17 and sim_3027 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_3027 >= 17 and sim_3027 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_3027 < 17 and sim_3027     == "y":
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan tidak boleh bawa motor")
print("Program Selesai")
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

<b>Penjelasan:</b><br><br>
<b>1. Mengambil Karakter Pertama dari Input SIM</b><br>
`sim_3027 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]`<br>
Masukan pengguna diambil karakter pertamanya sehingga program cukup membandingkan nilai awal tersebut dengan pilihan `'y'` atau `'t'`.

<br><br>
<b>2. Menguji Gabungan Dua Syarat</b><br>
`if umur_3027 >= 17 and sim_3027 == "y":`<br>
`elif umur_3027 >= 17 and sim_3027 != "y":`<br>
Operator `and` membuat kedua pemeriksaan harus sesuai agar suatu cabang dijalankan. Rangkaian `elif` kemudian menyediakan pemeriksaan berikutnya apabila kondisi sebelumnya tidak terpenuhi.

</div>

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px;">

### 2.2.5 Program Multi-IF 2 (`multi_if2_2611533027.py`)

</div>

````python
# Nama file program menggunakan format multi_if2_nim.py
# Program menerapkan percabangan kondisi
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Data pengguna dibaca melalui fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3027 = int(input("Input Total Belanja Anda = Masukkan total belanja (RP): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3027 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_3027 = input_member_3027 in ["y", "ya"]

# Input status promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3027 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_Valid_3027 = input_promo_3027 in ['y', 'ya']

total_diskon_persen_3027 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3027 > 100000:
    total_diskon_persen_3027 += 10  # Diskon belanja besar

if is_member_3027:
    total_diskon_persen_3027 += 5  # Diskon member

if kode_promo_Valid_3027:
    total_diskon_persen_3027 += 15  # Diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_3027 = total_belanja_3027 * (total_diskon_persen_3027 / 100)
total_bayar_3027 = total_belanja_3027 - nominal_diskon_3027

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon: {total_diskon_persen_3027}% (Rp {nominal_diskon_3027:,.0f})")
print(f"Total Bayar: Rp {total_bayar_3027:,.0f}")

print (f"Total diskon yang anda dapatkan adalah {total_diskon_persen_3027}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

<b>Penjelasan:</b><br><br>
<b>1. Membersihkan dan Memeriksa Input Member</b><br>
`is_member_3027 = input_member_3027 in ["y", "ya"]`<br>
Input terlebih dahulu dirapikan menggunakan `.strip()` dan `.lower()`. Setelah itu, nilai yang diperoleh diperiksa apakah termasuk pilihan `"y"` atau `"ya"` sehingga status member dapat disimpan sebagai nilai boolean.

<br><br>
<b>2. Menggabungkan Beberapa Potongan Diskon</b><br>
`if total_belanja_3027 > 100000:`<br>
&emsp;`total_diskon_persen_3027 += 10`<br>
`if is_member_3027:`<br>
&emsp;`total_diskon_persen_3027 += 5`<br>
`if kode_promo_Valid_3027:`<br>
&emsp;`total_diskon_persen_3027 += 15`<br>
Setiap aturan diskon menggunakan `if` sendiri. Dengan pola tersebut, seluruh potongan yang memenuhi syarat dapat dijumlahkan ke variabel total, bukan hanya mengambil salah satu kondisi.

</div>

### 2.2.7 Program Match Case 1 (`match_case1_2611533027.py`)

</div>

````python
# Nama file program menggunakan format match_case_nim.py
# Program menggunakan struktur match-case
# Penamaan variabel menggunakan empat digit terakhir NIM
# Data pengguna dibaca melalui fungsi input()
# Program mengubah nomor menjadi nama bulan

bulan_3027 = int(input("Inputkan nomor bulan (1-12): "))

match bulan_3027:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

<b>Penjelasan:</b><br><br>
<b>1. Menentukan Bulan Berdasarkan Angka</b><br>
`match bulan_3027:`<br>
Nilai yang diberikan pengguna dijadikan acuan untuk memilih salah satu `case` yang tersedia. Setiap angka dari 1 sampai 12 dipasangkan dengan nama bulan.

<br><br>
<b>2. Menangani Input di Luar Pilihan</b><br>
`case _:`<br>
Bagian ini berfungsi sebagai kondisi cadangan. Jika angka yang dimasukkan tidak sesuai dengan seluruh pilihan bulan, program menjalankan cabang tersebut dan menampilkan pesan bahwa input tidak valid.

````python
"""
Tugas 4 - Program Loket Wahana dan Perhitungan Transaksi
Nama   : Fairuz Ar Rasyd
NIM    : 2611533027
Materi : if tunggal, if-else, if-elif-else (operator logika), multi-if terpisah, match-case
"""

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. Membaca identitas dan data awal pengunjung
nama_3027 = input("Masukkan Nama Pengunjung        : ")
umur_3027 = int(input("Input umur anda                 : "))
sim_3027 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

print()
print("Pilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_3027 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_3027 = int(input("Masukkan jumlah tiket           : "))

# Pemeriksaan jumlah tiket menggunakan if tunggal
if jumlah_tiket_3027 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# 2. Menentukan wahana melalui match-case
nama_wahana_3027 = ""
harga_satuan_3027 = 0
transaksi_valid_3027 = True

match paket_3027:
    case 1:
        nama_wahana_3027 = "Wahana Safari Rimba"
        harga_satuan_3027 = 50000
    case 2:
        nama_wahana_3027 = "Wahana Arung Jeram"
        harga_satuan_3027 = 75000
    case 3:
        nama_wahana_3027 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3027 = 120000
    case 4:
        nama_wahana_3027 = "Wahana Roller Coaster Kilat"
        harga_satuan_3027 = 100000
    case 5:
        nama_wahana_3027 = "Wahana All-Access VIP"
        harga_satuan_3027 = 220000
    case _:
        print("Paket wahana tidak valid!")
        transaksi_valid_3027 = False

if transaksi_valid_3027:
    is_member_3027 = input("Apakah Anda member? (y/t)       : ").strip().lower()
    kode_promo_valid_3027 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

    print()
    print("--- KELAYAKAN PENGENDARA WAHANA ---")

    # 3. Memeriksa kelayakan pengunjung dengan beberapa kondisi
    if paket_3027 == 3:
        if umur_3027 >= 17 and sim_3027 == 'y':
            print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
        elif umur_3027 >= 17 and sim_3027 != 'y':
            print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_3027 < 17 and sim_3027 == 'y':
            print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
    else:
        # Paket selain ATV menggunakan pemeriksaan umur minimal 10 tahun
        if umur_3027 >= 10:
            print("Status Akses: Anda memenuhi syarat umur untuk menaiki wahana ini.")
        else:
            print("Status Akses: Anda belum memenuhi syarat umur untuk menaiki wahana ini.")

    # 4. Menggabungkan potongan melalui beberapa if
    subtotal_3027 = harga_satuan_3027 * jumlah_tiket_3027
    total_diskon_persen_3027 = 0

    if subtotal_3027 >= 200000:
        total_diskon_persen_3027 += 10  # Diskon Belanja Besar

    if is_member_3027 in ['y', 'ya']:
        total_diskon_persen_3027 += 5  # Diskon Member

    if kode_promo_valid_3027 in ['y', 'ya']:
        total_diskon_persen_3027 += 15  # Diskon Voucher Promo

    if jumlah_tiket_3027 >= 5:
        total_diskon_persen_3027 += 5  # Diskon Tambahan Rombongan

    # 5. Menentukan catatan layanan berdasarkan total pembayaran
    nominal_diskon_3027 = subtotal_3027 * (total_diskon_persen_3027 / 100)
    total_bayar_3027 = subtotal_3027 - nominal_diskon_3027

    if total_bayar_3027 > 300000:
        catatan_layanan_3027 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    else:
        catatan_layanan_3027 = "Terima kasih telah berkunjung."

    print()
    print("--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {subtotal_3027:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_3027}% (Rp {nominal_diskon_3027:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_3027:,.0f}")
    print(f"Catatan Layanan  : {catatan_layanan_3027}")
    print("Program Selesai")
````

`match-case` membandingkan nomor paket dengan pilihan yang tersedia. Lima pilihan pertama menentukan nama dan harga wahana, sedangkan `case _` menangani nomor yang tidak termasuk daftar. Pada kondisi tersebut status transaksi dibuat `False`.
