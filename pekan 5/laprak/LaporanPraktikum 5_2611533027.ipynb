## 2.1 Dasar Teori
### 1. Perulangan
Perulangan atau <i>looping</i> adalah mekanisme dalam pemrograman yang memungkinkan sebuah instruksi dijalankan berkali-kali tanpa harus menulis perintah yang sama secara berulang. Penggunaan perulangan membuat kode menjadi lebih ringkas dan membantu program menangani proses yang jumlahnya banyak. Pada Python, salah satu bentuk perulangan yang digunakan dalam praktikum adalah <i>for</i>, yaitu perulangan yang berjalan berdasarkan urutan atau rentang nilai tertentu.

### 2. Perulangan <i>for</i>
Perulangan <i>for</i> digunakan untuk mengakses setiap nilai dalam suatu urutan, termasuk hasil dari fungsi <i>range()</i>. Fungsi <i>range()</i> dapat menentukan nilai awal dan batas perulangan sehingga programmer dapat mengatur proses secara lebih terarah. Dalam praktikum, struktur ini digunakan untuk menampilkan angka, menghitung jumlah, dan mengulangi instruksi sesuai masukan pengguna.

### 3. Nested
<i>Nested</i> berarti suatu struktur pemrograman ditempatkan di dalam struktur lainnya. Penerapannya dapat berupa percabangan bersarang maupun perulangan bersarang. Dengan struktur tersebut, sebuah proses dapat memiliki beberapa tingkat pengendalian sehingga cocok digunakan untuk menyelesaikan persoalan yang membutuhkan pengolahan secara bertahap.

### 4. Nested Loop
<i>Nested loop</i> adalah perulangan yang berada di dalam perulangan lain. Perulangan bagian luar biasanya mengatur proses utama, sedangkan perulangan bagian dalam menjalankan proses untuk setiap bagian tersebut. Konsep ini sering digunakan untuk mengolah data berbentuk baris dan kolom serta membuat pola menggunakan karakter atau angka.

### 5. Penerapan Nested dan Perulangan
Nested dan perulangan dapat dikombinasikan untuk menghasilkan proses yang lebih terstruktur. Contohnya dapat dilihat ketika program membuat pola: perulangan luar menentukan baris, sedangkan perulangan dalam menentukan isi setiap posisi pada baris tersebut. Dengan cara ini, program dapat menghasilkan bentuk yang berulang dan teratur tanpa menuliskan setiap baris secara manual.

### 6. Manfaat Nested dan Perulangan
Pemahaman terhadap Nested dan Perulangan membantu mahasiswa menyusun algoritma yang lebih efisien dan memahami alur eksekusi program. Konsep ini juga melatih kemampuan dalam menentukan kondisi, batas proses, serta hubungan antara variabel dan hasil yang ditampilkan. Penguasaan dasar tersebut menjadi bekal untuk mempelajari pengolahan data dan struktur program yang lebih kompleks.

### 2.2.7 Program "Nested_For4_2611533027"
</div>

```python
#buat program perulangan for dalam python
#program ini menggunakan fungsi input

tinggi_3027 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3027 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3027 = tinggi_3027 
    c_3027 = a_3027
    lebar_3027 = (2 * tinggi_3027) - 2

    for i_3027 in range(1, tinggi_3027 + 1):
        b_3027 = c_3027 + 1

        for j_3027 in range(1, lebar_3027 + 1):

            # baris atas dan bawah

            if i_3027 == 1 or i_3027 == tinggi_3027:
                if j_3027 == 1 or j_3027 == lebar_3027:
                    print("#", end= " ")
                else:
                    print("=", end= " ")
            # Baris isi
            else:
                if j_3027 == 1 or j_3027 == lebar_3027:
                    print("|", end=" ")
                else:
                    if j_3027 == c_3027:
                        print("<", end="")
                    elif j_3027 == b_3027:
                        print(">", end=" ")
                    elif j_3027 == (lebar_3027 - c_3027):
                        print("<", end=" ")
                    elif j_3027 == (lebar_3027 - c_3027 + 1):
                        print(">", end=" ")
                    elif j_3027 > b_3027 and j_3027 < (lebar_3027 - c_3027):
                        print(".", end=" ")
                    else:
                        print(" ", end=" ")

        print()

        # logika asli java

        a_3027 -= 2

        if a_3027 <= 0:
            c_3027 = (-a_3027) + 2
        else:
            c_3027 = a_3027
```

<div style="font-family:'Times New Roman', Times, serif; font-size:18px;">
Penjelasan:<br><br>
1. Program meminta tinggi pola dan memeriksa apakah nilai tersebut merupakan bilangan genap menggunakan operator modulus.<br>
2. Jika nilai tidak genap, program memberikan pesan peringatan dan tidak melanjutkan pembentukan pola.<br>
3. Jika syarat terpenuhi, beberapa variabel seperti <i>a_3027</i>, <i>b_3027</i>, dan <i>c_3027</i> digunakan untuk menentukan posisi karakter.<br>
4. Perulangan <i>i_3027</i> mengatur baris, sedangkan <i>j_3027</i> mengatur posisi pada setiap baris. Keduanya membentuk nested loop.<br>
5. Kondisi <i>if</i> dan <i>elif</i> menentukan karakter yang dicetak berdasarkan posisi, sehingga bagian tepi dan bagian dalam pola dapat dibedakan.<br>
6. Setelah satu baris selesai, nilai posisi diperbarui sehingga susunan karakter berubah pada baris berikutnya.

<p style="text-indent: 40px; text-align: justify; "> Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan merupakan salah satu konsep penting dalam pemrograman Python yang digunakan untuk menjalankan suatu proses secara berulang sesuai dengan kondisi atau batas yang telah ditentukan. Perulangan for dapat digunakan untuk mengulang proses berdasarkan nilai yang terdapat pada suatu rentang tertentu. Pada praktikum, penggunaan input memungkinkan jumlah perulangan ditentukan secara langsung oleh pengguna sehingga program menjadi lebih fleksibel.<br>

Pada contoh program perulangan for, pengguna dapat menentukan jumlah perulangan, menampilkan urutan angka, serta melakukan proses penjumlahan dari angka yang dihasilkan. Penggunaan fungsi range() juga dapat mengatur nilai awal, nilai akhir, dan arah perulangan. Hal ini terlihat pada program yang menampilkan angka dari 0 sampai batas tertentu, angka dari 1 sampai batas tertentu, serta program yang menghitung jumlah dari seluruh angka dalam suatu rentang.<br>

Selain perulangan biasa, nested loop atau perulangan bersarang digunakan ketika terdapat perulangan di dalam perulangan lainnya. Nested loop dapat digunakan untuk membuat pola berdasarkan baris dan kolom. Pada praktikum, nested loop digunakan untuk membuat pola tanda bintang, pola angka, serta mengatur jumlah spasi dan karakter pada setiap baris.
